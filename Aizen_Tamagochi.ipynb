```python
import random
import time

# Status inicial do Aizen

fome = 70
felicidade = 70
energia = 70

# funções do bichinho

def barra(valor):
    cheios = int(valor / 10)
    vazios = 10 - cheios

    return " " * cheios + " " * vazios


def mostrar_status():
    print("\n" + "=" * 35)
    print("           ☽ AIZEN ☾")
    print("=" * 35)

    print(f"Fome:           {barra(fome)} {fome}/100")
    print(f"Felicidade:     {barra(felicidade)} {felicidade}/100")
    print(f"Energia:        {barra(energia)} {energia}/100")

    print("=" * 35)


def falar(frases):
    print("\nAizen:")
    print(f'"{random.choice(frases)}"')


def alimentar():
    global fome, felicidade

    fome = min(100, fome + 20)
    felicidade = min(100, felicidade + 10)

    falar([
        "Hm, aceitável...",
        "Você acha que pode me comprar com comida?",
        "Interessante",
        "Eu estava começando a sentir fome",
        "Continue assim"
    ])


def carinho():
    global felicidade, energia

    felicidade = min(100, felicidade + 20)
    energia = min(100, energia + 10)

    falar([
        "Não se apegue demais",
        "Você é bastante persistente",
        "Hm... Isso foi agradável",
        "Continue",
        "Que curioso"
    ])


def especial():
    global felicidade, energia

    item = input("\nO que você quer dar para Aizen? ").lower()

    if item == "livro":
        felicidade = min(100, felicidade + 20)
        energia = min(100, energia + 10)

        print("\n📖 Você deu um livro para Aizen")
        print("Interessante, uma leitura agradável")

    elif item == "chá":
        felicidade = min(100, felicidade + 20)
        energia = min(100, energia + 10)

        print("\n🍵 Você deu chá para Aizen.")
        print('"Uma escolha bastante apropriada."')

    elif item == "ovo cozido":
        felicidade = max(0, felicidade - 20)
        energia = max(0, energia - 10)

        print("\n🥚 Você deu um ovo cozido para Aizen.")
        print('"O que... é isso?"')

    else:
        print("\nAizen não parece interessado nisso.")


def passear():
    global fome, felicidade, energia

    if energia < 15:
        falar([
            "Estou sem energia",
            "Até mesmo eu preciso descansar",
            "Talvez devessemos ficar por aqui"
        ])
        return

    energia = max(0, energia - 20)
    fome = max(0, fome - 10)
    felicidade = min(100, felicidade + 20)

    falar([
        "Vamos observar o mundo",
        "Uma caminhada pode ser interessante",
        "Não se afaste de mim",
        "Vamos...",
        "Talvez encontremos algo interessante"
    ])


def passagem_tempo():
    global fome, felicidade, energia

    fome = max(0, fome - 2)
    felicidade = max(0, felicidade - 1)

    # Energia se recupera lentamente
    if energia < 100:
        energia = min(100, energia + 1)


# Menu

def menu():
    print("\n" + "=" * 35)
    print("           ☽ AIZEN ☾")
    print("=" * 35)

    print("1 - 🍙 Alimentar")
    print("2 - ❤️ Dar carinho")
    print("3 - 🎁 Dar item especial")
    print("4 - 🚶 Passear")
    print("5 - 📊 Ver status")
    print("6 - ❌ Sair")

    print("=" * 35)


# Inicio do jogo

print("\n" + "=" * 35)
print("           ☽ AIZEN ☾")
print("=" * 35)

print("\nAizen olha para você")

falar([
    "Oh, você veio me visitar",
    "Então você deciciu cuidar de mim",
    "Interessante... vamos ver o quanto isso dura",
    "Muito bem, pode começar"
])


# Loop principal

while True:
    menu()

    escolha = input("Escolha uma ação: ")

    if escolha == "1":
        alimentar()

    elif escolha == "2":
        carinho()

    elif escolha == "3":
        especial()

    elif escolha == "4":
        passear()

    elif escolha == "5":
        mostrar_status()

    elif escolha == "6":
        print("\nAté a próxima")
        break

    else:
        print("\nOpção inválida")

    passagem_tempo()

    time.sleep(1)
```
